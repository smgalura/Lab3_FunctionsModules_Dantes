```python
LAST_NAME = "DANTES"
SEED_NUM = 5
FAVORITE_ARTIST = "AL JAMES"

execution_log = []


# Decorator
def record(func):
    def wrapper(*args):
        try:
            result = func(*args)
            execution_log.append(func.__name__ + ": SUCCESS")
            return result
        except Exception as e:
            execution_log.append(func.__name__ + ": FAILED")
            raise
    return wrapper


# Generate equipment readings
@record
def generate_readings(name, seed, artist):
    name_sum = sum(ord(c) for c in name)
    artist_sum = sum(ord(c) for c in artist)

    return {
        "Temperature": 40 + (name_sum % 50) + seed,
        "Pressure": 20 + (artist_sum % 60) + seed,
        "Vibration": round(1 + ((name_sum + artist_sum) % 100) / 10, 1),
        "Voltage": 200 + ((name_sum * seed) % 40)
    }


# Validate readings
@record
def validate(readings):
    limits = {
        "Temperature": (0, 150),
        "Pressure": (0, 150),
        "Vibration": (0, 20),
        "Voltage": (100, 300)
    }

    results = {}

    for key, value in readings.items():
        low, high = limits[key]

        if not isinstance(value, (int, float)):
            raise TypeError(key + " must be numeric")

        if not low <= value <= high:
            raise ValueError(key + " is out of range")

        results[key] = "VALID"

    return results


# Calculate results
@record
def calculate(readings):
    values = list(readings.values())

    return {
        "Total": round(sum(values), 2),
        "Average": round(sum(values) / len(values), 2),
        "Highest": max(values),
        "Lowest": min(values)
    }


# Classify equipment
@record
def classify(stats):
    if stats["Average"] < 70:
        return "NORMAL"
    elif stats["Average"] < 120:
        return "WARNING"
    else:
        return "CRITICAL"


# Main program
try:
    readings = generate_readings(
        LAST_NAME,
        SEED_NUM,
        FAVORITE_ARTIST
    )

    validation = validate(readings)
    stats = calculate(readings)
    status = classify(stats)

    print("=== Generated Equipment Data ===")
    for key, value in readings.items():
        print(key + ":", value)

    print("\n=== Validation Results ===")
    for key, value in validation.items():
        print(key + ":", value)

    print("\n=== Diagnostic Results ===")
    for key, value in stats.items():
        print(key + ":", value)
    print("Status:", status)

    print("\n=== Execution Log ===")
    for log in execution_log:
        print(log)

except (ValueError, TypeError) as error:
    print("Diagnostic stopped:", error)
```

```python
# Recursive Fault Trace

def generate_fault_code():
    name_sum = sum(ord(c) for c in LAST_NAME)
    artist_sum = sum(ord(c) for c in FAVORITE_ARTIST if c != " ")

    fault_code = name_sum * (SEED_NUM + 1) + artist_sum

    execution_log.append("generate_fault_code: SUCCESS")

    return fault_code


def trace_fault(code, calls=1, trace=None):
    if trace is None:
        trace = []

    # Base condition
    if code < 10:
        trace.append(
            "Call " + str(calls) + ": " +
            str(code) + " -> ROOT FAULT"
        )
        execution_log.append("trace_fault: ROOT FAULT")
        return code, calls, trace

    next_code = sum(int(digit) for digit in str(code))

    trace.append(
        "Call " + str(calls) + ": " +
        str(code) + " -> " + str(next_code)
    )

    execution_log.append(
        "trace_fault: Call " + str(calls)
    )

    return trace_fault(next_code, calls + 1, trace)


# Generate fault code using values from the first program
fault_code = generate_fault_code()

root, calls, trace = trace_fault(fault_code)


# Output
print("=== Generated Fault Data ===")
print("Surname:", LAST_NAME)
print("SEED_NUM:", SEED_NUM)
print("Favorite Artist:", FAVORITE_ARTIST)
print("Fault Code:", fault_code)

print("\n=== Recursive Trace ===")
for step in trace:
    print(step)

print("\n=== Number of Recursive Calls ===")
print(calls)

print("\n=== Execution Log ===")
for log in execution_log:
    print(log)

print("\n=== Final Output ===")
print("Fault Code:", fault_code)
print("Root Fault:", root)
print("Recursive Calls:", calls)
```
